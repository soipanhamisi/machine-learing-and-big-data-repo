```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

data = load_breast_cancer()
breastCancer_df = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
data.keys()
```

```python
breastCancer_df['target'] = data.target
```

```python
breastCancer_df
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(breastCancer_df.drop('target', axis=1), breastCancer_df['target'], random_state=92)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.neural_network import MLPClassifier
breastCancer_mlpc = MLPClassifier().fit(X_train, y_train)
```

```python
breastCancer_mlpc1 = MLPClassifier(
    hidden_layer_sizes=[50, 100, 150, 200, 250, 200, 150, 100, 50, 25],
    alpha=0.1
).fit(X_train, y_train)
```

```python
print(f"Test set accuracy: {breastCancer_mlpc.score(X_test, y_test)}")
```

```python
print(f"Training set accuracy: {breastCancer_mlpc.score(X_train, y_train)}")
```

## feature scaling is important in ANNs

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
breastCancer_mlpc = MLPClassifier(max_iter=500).fit(X_train, y_train)
```

```python
print(f"Training set accuracy: {breastCancer_mlpc.score(X_train, y_train)}")
print(f"Test Set accuracy: {breastCancer_mlpc.score(X_test, y_test)}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(20, 5))
plt.imshow(breastCancer_mlpc.coefs_[0], interpolation='none', cmap='viridis')
plt.xticks(np.arange(len(breastCancer_df.columns)), breastCancer_df.columns)
plt.ylabel("Columns in weight matrix")
plt.xlabel("Input feature")
plt.colorbar()
```
